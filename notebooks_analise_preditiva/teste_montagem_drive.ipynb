# Teste de Montagem do Google Drive

Este notebook verifica se o Google Drive foi montado corretamente e se os arquivos estão acessíveis.

```python
# ============================================================================
# MONTAR GOOGLE DRIVE
# ============================================================================

try:
    from google.colab import drive
    drive.mount('/content/drive')
    print("✓ Google Drive montado com sucesso!")
except Exception as e:
    print(f"⚠️  Erro ao montar Google Drive: {e}")
```

```python
# ============================================================================
# VERIFICAR DIRETÓRIO dados_analise
# ============================================================================

import os

caminho_pasta = '/content/drive/MyDrive/dados_analise'

if os.path.exists(caminho_pasta):
    print(f"✓ Pasta encontrada: {caminho_pasta}")
    print(f"\nArquivos na pasta:")
    arquivos = os.listdir(caminho_pasta)
    for i, arquivo in enumerate(arquivos, 1):
        print(f"  {i}. {arquivo}")
else:
    print(f"⚠️  Pasta não encontrada: {caminho_pasta}")
    print("\nVerifique se você criou a pasta 'dados_analise' no Google Drive.")
```

```python
# ============================================================================
# TESTAR LEITURA DOS ARQUIVOS
# ============================================================================

import pandas as pd

# Lista de arquivos para testar
arquivos_para_testar = [
    'dataset_preditivo_com_precos.parquet',
    'ranking_risco_desmatamento_2023.parquet',
    'ranking_risco_embargos_2023.parquet'
]

for arquivo in arquivos_para_testar:
    caminho_completo = os.path.join(caminho_pasta, arquivo)
    print(f"\nTestando: {arquivo}")
    print(f"Caminho: {caminho_completo}")
    
    if os.path.exists(caminho_completo):
        try:
            df = pd.read_parquet(caminho_completo)
            print(f"✓ Arquivo lido com sucesso!")
            print(f"  Shape: {df.shape}")
            print(f"  Colunas: {list(df.columns[:5])}...")
        except Exception as e:
            print(f"⚠️  Erro ao ler arquivo: {e}")
    else:
        print(f"⚠️  Arquivo não encontrado")
```
