## Análise 1: Correlação entre Crescimento Populacional e Desmatamento

### Especificação BDD

**Feature**: Avaliar se crescimento populacional impulsiona desmatamento

**Scenario**: Calcular correlação entre delta populacional e delta desmatamento

- **GIVEN** que tenho dados de população municipal por ano (Censo IBGE)
- **AND** que tenho dados de desmatamento (PRODES) por município e ano
- **WHEN** eu calculo o delta de população entre dois períodos (ex: 2010-2020)
- **AND** calculo o delta de desmatamento no mesmo período
- **AND** calculo a densidade populacional (população / área municipal)
- **AND** calculo a correlação de Pearson entre delta população e delta desmatamento
- **AND** calculo a correlação entre densidade populacional e desmatamento per capita
- **AND** aplico regressão múltipla para controlar por outras variáveis
- **THEN** devo obter o coeficiente de correlação entre população e desmatamento
- **AND** devo identificar se crescimento populacional é driver de desmatamento
- **AND** devo calcular o desmatamento per capita (ha desmatado por habitante)

### Matemática em Linguagem de Negócio

**1. Delta Populacional:**
- Delta População = População_final - População_inicial
- Taxa de crescimento = (População_final - População_inicial) / População_inicial × 100%
- Taxa positiva = população cresceu
- Taxa negativa = população diminuiu (êxodo rural)

**2. Densidade Populacional:**
- Densidade = População / Área Municipal (hab/km²)
- Densidade alta (> 100 hab/km²): área urbanizada/consolidada
- Densidade média (20-100 hab/km²): área de ocupação moderada
- Densidade baixa (< 20 hab/km²): área rural/esparsa

**3. Desmatamento Per Capita:**
- Desmatamento per capita = Delta Desmatamento / População_média
- Mede hectares desmatados por habitante
- Permite comparação entre municípios de tamanhos diferentes
- Alto per capita: cada habitante associado a muito desmatamento
- Baixo per capita: cada habitante associado a pouco desmatamento

**4. Correlação de Pearson:**
- r > 0.5: correlação forte (crescimento populacional → desmatamento)
- r 0.3-0.5: correlação moderada
- r < 0.3: correlação fraca (população não é driver principal)

**5. Interpretação de Negócio:**
- Correlação forte: pressão demográfica impulsiona desmatamento
- Correlação fraca: outros fatores (ex: economia) são mais importantes
- Desmatamento per capita alto: padrão de uso intensivo de recursos
- Desmatamento per capita baixo: padrão de uso extensivo ou eficiente

**6. Limitações:**
- Correlação não implica causalidade
- Pode haver variável omitida (ex: infraestrutura)
- Migração pode não ser capturada pelo Censo (dados pontuais)
- Área municipal pode incluir áreas protegidas (viés de medição)
