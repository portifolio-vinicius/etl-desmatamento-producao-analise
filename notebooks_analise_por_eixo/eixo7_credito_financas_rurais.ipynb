## Analise 1: Correlacao entre Credito Rural e Desmatamento

### Especificacao BDD

Feature: Avaliar se credito rural impulsiona desmatamento

Scenario: Calcular correlacao entre concessao de credito e desmatamento

- GIVEN que tenho dados de credito rural por municipio e ano
- AND que tenho dados de desmatamento (PRODES) por municipio e ano
- WHEN eu calculo o delta de credito rural entre dois periodos
- AND calculo o delta de desmatamento no mesmo periodo
- AND calculo a correlacao de Pearson entre delta credito e delta desmatamento
- THEN devo obter o coeficiente de correlacao entre credito e desmatamento
- AND devo identificar se credito rural e driver de desmatamento

### Matematica em Linguagem de Negocio

1. Delta de Credito Rural:
- Delta Credito = Credito_final - Credito_inicial
- Taxa de crescimento = (Credito_final - Credito_inicial) / Credito_inicial x 100%

2. Credito Per Capita:
- Credito per capita = Credito / Populacao
- Mede R$ de credito por habitante

3. Correlacao de Pearson:
- r > 0.5: correlacao forte (credito -> desmatamento)
- r < 0.3: correlacao fraca (credito nao e driver principal)

4. Interpretacao de Negocio:
- Correlacao forte: credito rural impulsiona desmatamento
- Correlacao fraca: outros fatores sao mais importantes

5. Limitacoes:
- Correlacao nao implica causalidade
- Pode haver variavel omitida (precos agricolas)
- Dados de credito podem nao capturar financiamento informal
